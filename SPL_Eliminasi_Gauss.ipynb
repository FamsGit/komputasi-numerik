````python

def eliminasi_gauss(A, b):
    n = len(A)

    if n == 0 or len(b) != n or any(len(baris) != n for baris in A):
        raise ValueError("Ukuran matriks A dan vektor b tidak sesuai.")

    matriks = [
        [float(nilai) for nilai in A[i]] + [float(b[i])]
        for i in range(n)
    ]

    epsilon = 1e-12

    # Eliminasi maju
    for k in range(n):
        pivot = max(range(k, n), key=lambda i: abs(matriks[i][k]))

        if abs(matriks[pivot][k]) < epsilon:
            raise ValueError("SPL tidak memiliki solusi tunggal.")

        # Tukar baris jika diperlukan
        matriks[k], matriks[pivot] = matriks[pivot], matriks[k]

        # Mengubah matriks menjadi segitiga atas
        for i in range(k + 1, n):
            faktor = matriks[i][k] / matriks[k][k]

            for j in range(k, n + 1):
                matriks[i][j] -= faktor * matriks[k][j]

            matriks[i][k] = 0.0

    # Substitusi mundur
    x = [0.0] * n

    for i in range(n - 1, -1, -1):
        jumlah = sum(matriks[i][j] * x[j] for j in range(i + 1, n))
        x[i] = (matriks[i][n] - jumlah) / matriks[i][i]

    return x


# Contoh SPL:
# 2x + y - z = 8
# -3x - y + 2z = -11
# -2x + y + 2z = -3

A = [
    [2, 1, -1],
    [-3, -1, 2],
    [-2, 1, 2]
]

b = [8, -11, -3]

hasil = eliminasi_gauss(A, b)

print("Hasil Penyelesaian SPL:")
for i, nilai in enumerate(hasil, start=1):
    print(f"x{i} = {nilai:.2f}")
````

## Algoritma Penyelesaian Sistem Persamaan Linear (SPL) dengan Metode Eliminasi Gauss

**Tujuan:** Menyelesaikan Sistem Persamaan Linear (SPL) menggunakan metode eliminasi Gauss.

**Prinsip metode:**

Metode eliminasi Gauss adalah metode penyelesaian SPL dengan mengubah matriks koefisien menjadi matriks segitiga atas melalui operasi baris elementer. Setelah itu, nilai setiap variabel dihitung menggunakan substitusi mundur (*back substitution*).

**Pseudocode:**

```text
ALGORITMA ELIMINASI GAUSS

Input:
    Matriks koefisien A berukuran n × n
    Vektor konstanta b berukuran n

Proses:
    1. Gabungkan matriks A dan vektor b menjadi matriks augmented [A|b].

    2. Untuk setiap kolom k dari 1 sampai n:
       a. Pilih baris pivot dengan nilai absolut koefisien
          terbesar pada kolom k, mulai dari baris k.
       b. Tukar baris pivot dengan baris k jika diperlukan.
       c. Jika elemen pivot bernilai nol, hentikan proses
          karena solusi tunggal tidak dapat ditentukan.
       d. Untuk setiap baris i di bawah baris k:
          - Hitung faktor = A[i,k] / A[k,k].
          - Kurangi baris i dengan faktor dikalikan baris k.
          - Hasilkan elemen di bawah pivot menjadi nol.

    3. Lakukan substitusi mundur dari baris terakhir
       hingga baris pertama untuk menghitung nilai variabel.

Output:
    Nilai setiap variabel sebagai solusi SPL.
</text>
```

**Catatan**: Pemilihan pivot dengan nilai absolut terbesar merupakan teknik partial pivoting untuk membantu mengurangi masalah numerik saat perhitungan.

# **ANALISIS KODE PHYTON**

###**Bagian 1 — Memasukkan Data SPL**

Tujuan: Menyimpan koefisien setiap variabel dan nilai konstanta dari sistem persamaan linear ke dalam Python.

````python

A = [
    [2, 1, -1],
    [-3, -1, 2],
    [-2, 1, 2]
]

b = [8, -11, -3]

print("Matriks koefisien A:")
for baris in A:
    print(baris)

print("Vektor konstanta b:", b)
````

###**Bagian 2 — Membentuk Matriks Augmented**

Tujuan: Menggabungkan matriks koefisien A dengan vektor konstanta b menjadi satu matriks agar operasi eliminasi lebih mudah dilakukan.

````python

def buat_matriks_augmented(A, b):
    n = len(A)

    if n == 0:
        raise ValueError("Matriks A tidak boleh kosong.")

    if len(b) != n or any(len(baris) != n for baris in A):
        raise ValueError("Ukuran matriks A dan vektor b tidak sesuai.")

    matriks_augmented = []

    for i in range(n):
        baris = [float(nilai) for nilai in A[i]]
        baris.append(float(b[i]))
        matriks_augmented.append(baris)

    return matriks_augmented
````

###**Bagian 3 — Melakukan eliminasi maju (forward elimination)**

Tujuan: Mengubah matriks augmented menjadi matriks segitiga atas dengan membuat elemen di bawah elemen pivot menjadi nol.

````python

def eliminasi_maju(matriks):
    n = len(matriks)
    epsilon = 1e-12

    for k in range(n):
        pivot = max(
            range(k, n),
            key=lambda i: abs(matriks[i][k])
        )

        if abs(matriks[pivot][k]) < epsilon:
            raise ValueError(
                "Matriks singular; kode ini hanya menyelesaikan "
                "SPL yang memiliki solusi tunggal."
            )

        matriks[k], matriks[pivot] = matriks[pivot], matriks[k]

        for i in range(k + 1, n):
            faktor = matriks[i][k] / matriks[k][k]

            for j in range(k, n + 1):
                matriks[i][j] -= faktor * matriks[k][j]

            matriks[i][k] = 0.0

    return matriks
````

###**Bagian 4 — Melakukan Substitusi Mundur (Back Substitution)**

Tujuan: Menghitung nilai setiap variabel setelah matriks menjadi berbentuk segitiga atas.

````python

def substitusi_mundur(matriks):
    n = len(matriks)
    epsilon = 1e-12

    x = [0.0] * n

    for i in range(n - 1, -1, -1):
        if abs(matriks[i][i]) < epsilon:
            raise ValueError(
                "Elemen diagonal terlalu dekat dengan nol."
            )

        jumlah = sum(
            matriks[i][j] * x[j]
            for j in range(i + 1, n)
        )

        x[i] = (matriks[i][n] - jumlah) / matriks[i][i]

    return x
````

### **Bagian 5 — Menggabungkan Tahapan Algoritma**

Tujuan: Membuat satu fungsi utama yang memanggil proses pembentukan matriks, eliminasi maju, dan substitusi mundur secara berurutan.

````python

def eliminasi_gauss(A, b):
    matriks = buat_matriks_augmented(A, b)
    matriks = eliminasi_maju(matriks)
    hasil = substitusi_mundur(matriks)

    return hasil
````

### **Bagian 6 — Menjalankan program dan menampilkan hasil**

Tujuan: Memasukkan data ke fungsi utama, menjalankan perhitungan, dan menampilkan nilai setiap variabel.

````python

nama_variabel = ["x", "y", "z"]

try:
    hasil = eliminasi_gauss(A, b)

    print("Hasil Penyelesaian SPL:")
    for nama, nilai in zip(nama_variabel, hasil):
        print(f"{nama} = {nilai:.2f}")

except ValueError as error:
    print("Terjadi kesalahan:", error)
````

### **Tambahan**
Cell ini menampilkan matriks sebelum dan sesudah eliminasi maju.

````python

print("Matriks augmented awal:")
matriks_awal = buat_matriks_augmented(A, b)

for baris in matriks_awal:
    print([round(nilai, 2) for nilai in baris])

print("\nMatriks setelah eliminasi maju:")
matriks_segita_atas = eliminasi_maju(
    buat_matriks_augmented(A, b)
)

for baris in matriks_segita_atas:
    print([round(nilai, 2) for nilai in baris])
````
